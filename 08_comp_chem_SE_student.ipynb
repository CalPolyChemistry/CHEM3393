## Background: The Schrödinger equation as a matrix equation

In today’s lab, you will be directly solving the Schrödinger equation in a matrix
equation form to determine the vibrational energy levels for carbon monoxide. As you have learned, the form of the Schrödinger equation is

$$
\mathbf{H}\mathbf{U} = \mathbf{E} \mathbf{U}
$$

In a matrix form, H is the Hamiltonian matrix, U is the wavefunction matrix (also called the eigenfunction or eigenvector matrix), and E is a diagonal matrix of the energies for each level n.
To solve this matrix equation for E, we must move U to the other side by taking the transpose of the matrix.

$$
\mathbf{U}^{T} \mathbf{H} \mathbf{U} = \mathbf{E}
$$


The Hamiltonian matrix contains kinetic and potential energy terms. The matrix must be chosen to be a finite size, and a grid of points to calculate the Hamiltonian must be constructed. You can choose any size for the Hamiltonian matrix that you wish, where more grid points would always give you a better answer. Optimizing the size of the Hamiltonian matrix is one possible part of the computational experiment, but for our purposes, we will choose to use a 800 by 800 matrix.
The kinetic energy elements of the Hamiltonian matrix are given by the equation

$$
\frac{\hslash^2 (-1)^{i-j}}{2 \mu \Delta r^2} A_{ij}
$$

where

$$
A_{ij} = \frac{2}{(i-j)^2} - \frac{2}{(i+j)^2}~\mathrm{for}~i \neq j
$$

or,

$$
A_{ij} = \frac{\pi^2}{3} - \frac{1}{2 i^2}~\mathrm{for}~i = j
$$

The potential energy elements of the Hamiltonian matrix are given by the equation

$$
V_{ij} = V(r_i) \delta_{ij}
$$


The $\delta_{ij}$ means that the potential contribution is 0 unless i = j. In our equation, we will use a Morse potential form for V.

$$
V(r_{i}) = D_e (1 - e^{-a(r_i - r_{eq})})^2
$$


In both equations, both i and j count over the grid elements. The grid points $r_i$ are uniformly spaced such that

$$
r_i = r_{min} + i \Delta r~\mathrm{for}~i=1,2,3,\dots,N
$$

where ∆r is the grid spacing and N is the number of grid points.

**NOTE!** Pay close attention to the bounds for the indicies i, and j above. The equations indicate that i, and j start at 1 (but do python lists/arrays start indexing at 1?!) No, no they do not, you will need to account for this.
