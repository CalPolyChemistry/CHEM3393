## Background: rotation of a rigid diatomic molecule

For a rigid diatomic molecule, the rotational energy levels are

$$E_J = h c \tilde{B}J(J+1), \qquad J=0,1,2,\ldots$$

where $\tilde{B}$ is the rotational constant, $h$ is Planck's constant, and $c$ is the speed of light. Note that the rotational constant is typically given in wavenumbers ($\mathrm{cm}^{-1}$).  

Each level has a degeneracy of

$$g_J = 2J+1.$$

For a diatomic molecule with equilibrium bond length $r_e$, the moment of inertia is

$$I=\mu r_e^2$$

where the reduced mass is

$$\mu=\frac{m_1m_2}{m_1+m_2}.$$

The rotational constant can be calculated from the moment of inertia:

$$\tilde{B}=\frac{h}{8\pi^2 c I}.$$

The rotational partition function is a Boltzmann-weighted sum over levels:

$$q_{\mathrm{rot}}=\sum_{J=0}^{\infty}g_J e^{-\beta E_J}=\sum_{J=0}^{\infty}(2J+1)\exp\left[-\frac{hc\tilde{B}J(J+1)}{k_BT}\right].$$

If many rotational levels are accessible, you can replace the summation with integration and derive a closed-form expression for the partition function:

$$q_{\mathrm{rot,closed}}=\frac{kT}{\sigma hc\tilde{B}}=\frac{T}{\sigma\theta_{\mathrm{rot}}},$$

where $\sigma$ is the rotational symmetry number and $\theta_{\mathrm{rot}}=hc\tilde{B}/k$ is the characteristic rotational temperature. For heteronuclear HCl, $\sigma=1$.
