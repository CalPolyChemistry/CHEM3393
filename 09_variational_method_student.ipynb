# Calculating the Energy of the Harmonic Oscillator using the Variational Method


## Background

The **variational method** is a numerical technique used to find parameters of a trial function that minimize the overall value of the function.  
In **quantum mechanics**, the variational method is often used to find the coefficients and energies of molecular orbitals. 


### The Variational Method
To use the variational method, a **trial function** $\phi$ is usually given. This function contains a **scalable parameter**.  
The variational method can then be used to find the value of this parameter which leads to the **lowest energy** of the trial function.  
Since the variational method always gives an **upper bound** to the real ground state energy, this yields the best estimate for the energy for the given trial function. 


### Energy of the Trial Function 
The energy of the trial function is given by the equation

$$
E_{\phi} = \frac{\int \phi^* \, \hat{H} \, \phi \, dx}{\int \phi^* \, \phi \, dx}.
$$
The denominator is required because it is not guaranteed that the trial function is normalized. 


### Minimizing the Energy 
Once an expression for the energy of the trial function is determined, we want to find the value of the scalable parameter that **minimizes the energy**.  
To do this, take the derivative of the energy expression with respect to the parameter, set the derivative equal to zero, and solve for the parameter:

$$
\frac{d E_{\phi}}{d\alpha} = 0.
$$
Solve for $\alpha$. Once you obtain the value of $\alpha$, substitute it back into the original energy expression for the trial function to get the **variational estimate of the energy**.
