# The Boltzmann Distribution and the Partition Function

In the previous lesson, random energy exchanges produced a predictable distribution of molecular energies. We can calculate that distribution directly using the **Boltzmann distribution** and the **partition function**.

At equilibrium, the Boltzmann distribution gives the fraction of molecules in energy level $\epsilon_i$:

$$p_I=\frac{g_I e^{-\beta\varepsilon_I}}{q},\qquad
\beta=\frac{1}{kT},\qquad q=\sum_I g_I e^{-\beta\varepsilon_I}$$

where $p_I$ is the fraction of molecules at the energy level $\epsilon_I$, $g_I$ is the degeneracy of the energy level (the number of states with the same energy), $T$ is the absolute temperature and $k$ is Boltzmann's constant. The partition function $q$ sums over energy levels, counting the $g_I$ states at each energy. When all levels are singly degenerate ($g_I=1$) these expressions reduce to 

$$p_i=\frac{e^{-\beta\varepsilon_i}}{q},\qquad q=\sum_i e^{-\beta\varepsilon_i}$$

At low temperature, most molecules occupy the lowest states. At higher temperature, $\beta$ is smaller and the populations decrease more slowly with energy: the distribution spreads across more states. The energy spacings matter as well as temperature. An energy level spacing much larger than $kT$ makes higher energy states difficult to populate.

The mean molecular energy can be determined from a weighted average of the populations of each energy level:

$$\langle\varepsilon\rangle=\sum_i p_i\varepsilon_i$$

## Part 3: CO Vibrational States

The energy of an individual molecule is the sum of its translational, rotational, vibrational, and electronic energies. The partition coefficients for each of these contributions can be calculated separately and then combined to give the total partition function for the molecule.

$$q = q^T q^R q^V q^E$$
